Vytvořte třídu Člověk, která bude mít atributy:


1.   jméno:str
2.   příjmení:str
3.   telefon:int
4.   email:str
5. adresa:str
6. pojišťovna:str

Vytvořte metody, které budou moc získávat a měnit tyto atributy. Také vytvořte metodu, která vám vypíše vše o této osobě




````python
class Clovek:
    def __init__(self, jmeno, prijmeni, tel, mail, adresa, pojistovna):
        self.jmeno = jmeno
        self.prijmeni = prijmeni
        self.tel = tel
        self.mail = mail
        self.adresa = adresa
        self.pojistovna = pojistovna

    def get_jmeno(self): #získá atribut. musí být napsáno pokud je atribut zkryt "__atribut"
        return self.jmeno
    def set_jmeno(self,nove_jmeno): #změní atribut
        self.jmeno = nove_jmeno
        print ("jmeno bylo změněno")

    def get_prijmeni(self):
        return self.prijmeni
    def set_prijmeni(self,nove_prijmeni):
        self.prijmeni = nove_prijmeni
        print ("příjmení bylo změněno")

    def get_tel(self):
        return self.tel
    def set_tel(self,nove_tel):
        self.tel = nove_tel
        print ("telefoní číslo bylo změněno")

    def get_mail(self):
        return self.mail
    def set_mail(self,nove_mail):
        self.mail = nove_mail
        print ("email byl změněn")

    def get_adresa(self):
        return self.adresa
    def set_adresa(self,nove_adresa):
        self.adresa = nove_adresa
        print ("adresa byla změněna")

    def get_pojistovna(self):
        return self.pojistovna
    def set_pojistovna(self,nove_pojistovna):
        self.pojistovna = nove_pojistovna
        print ("pojišťovna byla změněna")


    def __str__(self): #vypíše všechny atributy a informace třídy do stringu
        return f"""
        Jmeno: {self.jmeno}
        prijmeni: {self.prijmeni}
        tel: {self.tel}
        mail: {self.mail}
        adresa: {self.adresa}
        pojistovna: {self.pojistovna}
        """
#nad tímto je šablona

clovek1 = Clovek("aneta","černá",314720069,"aneta@gmail.com","šrekova bažina 13","VZP")
clovek2 = Clovek("karel","rudý",920476312,"rdy@gmail.com","pazkvylovy pysky 23","VZDR")
#nad tímto jsou údaje do šablony

print(clovek1.get_jmeno()) #získá pouze daný atribut (v tomto případě jméno)
print(clovek2)

clovek2.set_prijmeni(input("zadej nové příjmení")) #přepíše atribut

print(clovek1)
print(clovek2)
````

````python
class Pojistovna:
    def __init__(self, nazev:str,zkratka:str,sidlo:str):
        self.nazev = nazev
        self.zkratka = zkratka
        self.sidlo = sidlo
        self.setzkratka(zkratka)


    def getnazev(self):
        return self.nazev
    def setnazev(self,nazev):
        self.nazev = nazev

    def getzkratka(self):
        return self.zkratka
    def setzkratka(self,zkratka):
        if len(str(zkratka)) < 4 and len(str(zkratka)) > 0:
            self.zkratka = zkratka
        else:
            print("zadanná zkratka je neplatná")
            self.zkratka = ""

    def getsidlo(self):
        return self.sidlo
    def setsidlo(self,sidlo):
        self.sidlo = sidlo

    def __str__(self):
        return """
Název pojišťovny: {}
zkratka pojišťovny: {}
sídlo pojišťovny: {}
                """.format(self.nazev,self.zkratka,self.sidlo)
````

Vytvořte metody, které budou moc získávat a měnit tyto atributy. Také vytvořte metodu, která vám vypíše vše tomto počasí


````python
class Pocasi:
    def __init__(self,vitr:float,teplota:float,vlhkost:float,pocasi:str):
        self.vitr = vitr
        self.pocasi = pocasi
        self.teplota = teplota
        self.vlhkost = vlhkost

    def getvitr(self):
        return self.vitr
    def setvitr(self,vitr):
        self.vitr = vitr
    def getpocasi(self):
        return self.pocasi
    def setpocasi(self,pocasi):
        self.pocasi = pocasi
    def getteplota(self):
        return self.teplota
    def setteplota(self,teplota):
        self.teplota = teplota
    def getvlhkost(self):
        return self.vlhkost
    def setvlhkost(self,vlhkost):
        self.vlhkost = vlhkost

    def __str__(self):
        return """
Rychlost větru: {} km/h
Teplota: {} °C
Vlhkost vzduchu: {} %
Počasí: {}
""".format(self.vitr,self.teplota,self.vlhkost,self.pocasi)
````

# Počasí
Upravte třídu Pocasi, aby přijímala Enum (na výběr z: Slunečno, zataženo, bouřky atd.)


````python
from enum import Enum
class Pocasicko(Enum):
    SLUNECNO = "Slunečno"
    ZATAZENO = "Zataženo"
    BOURKY = "Bouřky"
    SNEZI = "sněží"

class Pocasi:
    def __init__(self,vitr:float,teplota:float,vlhkost:float,pocasi:Pocasicko):
        self.vitr = vitr
        self.pocasi = pocasi
        self.teplota = teplota
        self.vlhkost = vlhkost

    def getvitr(self):
        return self.vitr
    def setvitr(self,vitr):
        self.vitr = vitr
    def getpocasi(self):
        return self.pocasi
    def setpocasi(self,pocasi):
        self.pocasi = pocasi
    def getteplota(self):
        return self.teplota
    def setteplota(self,teplota):
        self.teplota = teplota
    def getvlhkost(self):
        return self.vlhkost
    def setvlhkost(self,vlhkost):
        self.vlhkost = vlhkost

    def __str__(self):
        return """
Rychlost větru: {} km/h
Teplota: {} °C
Vlhkost vzduchu: {} %
Počasí: {}
""".format(self.vitr,self.teplota,self.vlhkost,self.pocasi)
````

````python
pocasi3 = Pocasi(17,18,54,Pocasicko.SLUNECNO.value)
pocasi4 = Pocasi(22,32,84,Pocasicko.SNEZI.value)

print(pocasi1)
print(pocasi2)
print(pocasi3)
print(pocasi4)
````
