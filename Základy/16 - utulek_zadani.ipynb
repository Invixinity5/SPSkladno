Vytvořte třídu Utulek, který bude obsahovat tyto vlastnosti:

*   Název
*   Maximální kapacita
*   SeznamZvířat

A jeho metody budou:

*  getNazev
*  getMaxKapacita
*  getSeznamZvirat
*  setNazev
*  setMaxKapacita
*  pridatZvire
*  odeberZvire
*  ToString







```python
class Utulek:
    def __init__(self,nazev,max_cap:int,seznam=None):
        self.nazev = nazev
        self.max_cap = max_cap
        if seznam is None:
            self.seznam = "prázdno"
        else:
            self.seznam = seznam

    def pridej_zvire(self,zvire):
        if len(self.seznam) < self.max_cap:
            if zvire in self.seznam:
                print(f"{zvire} je již v útulku")
            else:
                self.seznam.append(zvire)
        else:
            print("útulek je plný")
    def odeber_zvire(self,zvire):
        if zvire in self.seznam:
            self.seznam.remove(zvire)
            print("zvíře bylo odebráno")
        else:
            print("zvíře není v útulku")
    def __str__(self):
        return """
Útulek: {}
kapacita: {}
seznam: {}
""".format(self.nazev,self.max_cap,self.seznam)

zver1 = "pes"
zver2 = "kočka"
utulek_bez_zvere = Utulek("hell",3,None)
utulek_se_zveri = Utulek("heaven's enigma",8,[zver1, zver2])

print(utulek_bez_zvere)
print(utulek_se_zveri)
```
